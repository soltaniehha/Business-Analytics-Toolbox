## Query with `%%bigquery`

A cell that starts with `%%bigquery` is a SQL cell. A name after it saves the result as a pandas DataFrame. This query finds the five longest Shakespeare works by word count.

```python
%%bigquery top_works
SELECT corpus, SUM(word_count) AS total_words
FROM `bigquery-public-data.samples.shakespeare`
GROUP BY corpus
ORDER BY total_words DESC
LIMIT 5
```

```python
top_works
```

`top_works` is a regular DataFrame, so pandas can chart it.

```python
top_works.plot(kind='barh', x='corpus', y='total_words', legend=False).invert_yaxis()
```
